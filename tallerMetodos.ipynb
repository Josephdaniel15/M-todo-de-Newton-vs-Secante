# Objetivo 1 - graficar la función
+ Vectores xs
+ Definir Función f(x)
+ Calcular vector f(xs) = ys


```python
import numpy as np

xs = list(range(-2, 6))
xs = list(np.linspace(-2, 6, 100))
def f(x):
    _y = x**3 - 3 * x**2 + x - 1
    return _y
ys = [f(x) for x in xs]
```

```python
import matplotlib.pyplot as plt

plt.plot(xs, ys, ".:")
plt.xlabel("xs")
plt.ylabel("ys")
plt.title("Gráfica de ys vs xs")
plt.grid(True)
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.optimize import newton
def f(x): return x**3 - 3*x**2 + x
def df(x): return 3*x**2 - 6*x + 1 
xs_puntos = []
ys_puntos = []
def f_guardar(x):
    y = f(x)
    xs_puntos.append(x)
    ys_puntos.append(y)
    return y
xn = newton(f_guardar, x0 = 100, fprime=df)
xs = np.array(xs_puntos)
ys = np.array(ys_puntos)
plt.plot(xs, ys, ".:")
for i in range(len(xs)):
    plt.plot(xs[i], ys[i], "ro")
plt.xlabel("xs")
plt.ylabel("ys")
plt.title("Gráfica de ys vs xs")
plt.grid(True)
plt.show()
```

```python
import matplotlib.pyplot as plt
from scipy.optimize import newton
import numpy as np
def f(x):
    return x**3 - 3*x**2 + x + 1
def df(x):
    return 3*x**2 - 6*x + 1
xn = newton(f, x0=100, fprime=df)
xs = np.linspace(-2, 5, 200)
ys = f(xs)
plt.plot(xs, ys, ".:")
plt.plot(xn, f(xn), "ro", label=f"Raíz ≈ {xn:.4f}")
plt.xlabel("xs")
plt.ylabel("ys")
plt.title("Gráfica de ys vs xs")
plt.grid(True)
plt.legend()
plt.show()
```

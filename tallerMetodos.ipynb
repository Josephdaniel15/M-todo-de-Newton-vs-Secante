```python
# alternativa 2
from scipy.optimize import newton
import numpy as np
import matplotlib.pyplot as plt

def f(x):
    _y = x**3 - 3 * x**2 + x - 1
    print(f"{x:.2f} \t {_y:.2f}")
    return _y

def df(x):
    _dy = 3 * x**2 - 6 * x + 1
    # print(f"{x:.2f} \t {_dy:.2f}")
    return _dy

xs = list(np.linspace(-1, 5, 100))
ys = [f(x) for x in xs]

plt.plot(xs, ys)
plt.xlabel("xs")
plt.ylabel("ys")
plt.title("Gráfica de ys vs xs")
plt.grid(True)
plt.show()

xn = newton(f, x0=100, fprime=df)
xn
```

```python
xn_iter = [-10.0]

for i in range(10000):
    x = xn_iter[-1]
    x_next = x - f(x)/df(x)
    xn_iter.append(x_next)

plt.figure(figsize=(10, 6))
plt.plot(range(len(xn_iter)), xn_iter, 'bo-', linewidth=2, markersize=6)
plt.xlabel('Iteración')
plt.ylabel('$x_n$')
plt.title('Convergencia del método de Newton')
plt.grid(True)

plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt

def f(x):
    return x**3 - 3 * x**2 + x - 1

def df(x):
    return 3 * x**2 - 6 * x + 1

x0_vals = [-10.0, 0.0]

plt.figure(figsize=(12,7))

x = np.linspace(-11, 1, 800)
plt.plot(x, f(x), color='black', label='f(x)')

for x0 in x0_vals:

    xn_iter = [x0]

    for i in range(5):
        xk = xn_iter[-1]
        x_next = xk - f(xk)/df(xk)
        xn_iter.append(x_next)

    for i in range(len(xn_iter)-1):

        xi = xn_iter[i]
        yi = f(xi)
        pendiente = df(xi)

        xt = np.linspace(xi-1.5, xi+1.5, 100)
        yt = yi + pendiente*(xt - xi)

        if x0 == -10.0:
            plt.plot(xt, yt, color='blue', label='Convergente' if i == 0 else "")
            plt.plot(xi, yi, 'bo')
            plt.plot(xn_iter[i+1], 0, 'bo')

        else:
            plt.plot(xt, yt, color='red', linestyle='--', label='Divergente' if i == 0 else "")
            plt.plot(xi, yi, 'ro')
            plt.plot(xn_iter[i+1], 0, 'ro')

plt.axhline(0, color='black')

plt.xlim(-7, 2)
plt.ylim(-400, 50)

plt.xlabel('x')
plt.ylabel('f(x)')
plt.title('Método de Newton con tangentes')
plt.grid(True)
plt.legend()

plt.show()
```
